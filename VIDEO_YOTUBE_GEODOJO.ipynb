```python
escuelas_privadas_capital = escuelas_privadas[escuelas_privadas['est_depart'] == 'CAPITAL']
display(escuelas_privadas_capital.head())
display(escuelas_privadas_capital.shape)
```

```python
# Leer archivo shp
departamentos = gpd.read_file("/content/drive/MyDrive/VIDEOS_YOTUBE/DATOS/departamentos.shp")
```

```python
departamentos.head()
```

```python
departamentos.shape
```

```python
import matplotlib.pyplot as plt
departamentos.plot(figsize=(12,10), legend=True, column="nombre", legend_kwds={'loc': 'upper left', 'bbox_to_anchor': (1, 1)})
plt.tight_layout()
plt.show()
```

```python
# Cortar el poligono capital
departamentos_cordoba = departamentos[departamentos["nombre"] == "CAPITAL"]
```

```python
departamentos_cordoba.plot(figsize=(12,10), legend=True, column="nombre", legend_kwds={'loc': 'upper left', 'bbox_to_anchor': (1, 1)})
plt.tight_layout()
plt.show()
```

```python
# 1. Definir el CRS específico para POSGAR 2007 / Argentina Faja 3 (EPSG:5345)
crs_especifico = "EPSG:5345"

# 2. Reproyectar ambos GeoDataFrames al mismo CRS métrico proyectado
escuelas_reproyectadas_esp = escuelas_privadas_capital.to_crs(crs_especifico)
departamentos_reproyectados_esp = departamentos_cordoba.to_crs(crs_especifico)

# 3. Realizar el cruce espacial (Spatial Join) con el nuevo CRS
escuelas_en_capital_especifico = gpd.sjoin(
    escuelas_reproyectadas_esp,
    departamentos_reproyectados_esp,
    how="inner",
    predicate="within"
)

# Mostrar el resultado y verificar el CRS
print(f"CRS actual de los datos: {escuelas_en_capital_especifico.crs}")
print(f"Cantidad de escuelas encontradas: {len(escuelas_en_capital_especifico)}")
display(escuelas_en_capital_especifico.head())
```

```python
# 3. Graficar el resultado para verificar visualmente (ambos capas en EPSG:5345)
fig, ax = plt.subplots(figsize=(10, 10))

# Graficamos el departamento reproyectado de fondo
departamentos_reproyectados_esp.plot(ax=ax, color='lightgrey', edgecolor='black', alpha=0.6, label='Departamento Capital')

# Graficamos las escuelas reproyectadas sobre el departamento
escuelas_en_capital_especifico.plot(ax=ax, color='red', markersize=25, alpha=0.8, label='Escuelas Privadas')

plt.title('Escuelas Privadas dentro del Departamento Capital\n(POSGAR 2007 / Argentina Faja 3)', fontsize=14)
plt.xlabel('Este [metros]', fontsize=10)
plt.ylabel('Norte [metros]', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
# Guardar ambos archivos en geopakage
departamentos_cordoba.to_file("/content/departamentos_cordoba.gpkg", layer='departamentos_cordoba', driver="GPKG")
escuelas_privadas_capital.to_file("/content/escuelas_privadas_capital.gpkg", layer='escuelas_privadas_capital', driver="GPKG")
```
